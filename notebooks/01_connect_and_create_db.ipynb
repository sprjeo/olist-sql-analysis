```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine, text
```

```python
load_dotenv()
```

```python
user = os.getenv("DB_USER")
password = os.getenv("DB_PASSWORD")
host = os.getenv("DB_HOST")
port = os.getenv("DB_PORT")
db = os.getenv("DB_NAME")
```

```python
url_autocommit = f"postgresql://{user}:{password}@{host}:{port}/{db}"

engine_autocommit = create_engine(url_autocommit, isolation_level="AUTOCOMMIT")

with engine_autocommit.connect() as conn:
    conn.execute(text("CREATE DATABASE olist;"))
```

```python
url_olist = f"postgresql://{user}:{password}@{host}:{port}/olist"
engine_olist = create_engine(url_olist)

with engine_olist.connect() as conn:
    result = conn.execute(text("SELECT current_database();"))
    print(result.fetchone())
```
